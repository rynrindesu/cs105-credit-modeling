### 2. Data pre-processing [35% of Part I]

**a.** For each variable, determine the percentage of missing data. For any column with missing data, describe how you resolve the issue. Clearly state any assumption you made.

**Response.** 

```python
# Calculate the percentage of valid data
percentage_missing = 1 - data.count() / rows

# Filter for columns with missing data
percentage_missing[percentage_missing > 0.0]

print(f" Number of variables with missing data: {len(percentage_missing[percentage_missing > 0.0])}")
print(f"")

# Display the variables with missing data 
missing_data_df = percentage_missing[percentage_missing > 0.0].reset_index()
missing_data_df.columns = ['Field', '% of Missing Data']

# Display the result
print(missing_data_df)
```

```python
# Display the percentage of missing data for these 2 fields

percentage_missing[percentage_missing>0.0].hist()
plt.xlabel("% of missing data")
plt.ylabel("# of variables")
plt.show()
```

The percentage of missing data is insignificant compared to the amount of data we have, both variables have around `5% - 6%` of missing data.
Since we know both `Checking Account` and `Dependents` are categorical data, we will impute it with the mode category.

```python
# Get the two variables and impute the missing fields with the mode

threshold = 0.10
impute_cat = percentage_missing[(percentage_missing<threshold)&(percentage_missing>0.0)].index
impute_cat_mode = list(set(impute_cat))
print(f"Categorical columns to impute with mode: {impute_cat_mode}\n")

for col in impute_cat_mode:
    data[col].fillna(data[col].mode().iloc[0], inplace=True)

# Confirm the variables do not have any NaN values 

print(f"After imputation: ")
print(data[impute_cat_mode].count())
```

**b.** For each variable, identify outliers (if any) and describe how you resolve the issue. Clearly state any assumption you made.

**Response.** 

```python
import seaborn as sns

# Certain numerical variables selected to check for outliers
ranged_data = data[['Amount', 'Installment', 'Age', 'Num_Credits']]

# Create a boxplot for each of the variables 
fig, axes = plt.subplots(1, 4, figsize=(8, 5))  
axes = axes.flatten()  

# Iterates over ranged_data to create the boxplot
for i, column in enumerate(ranged_data.columns):
    sns.boxplot(data=ranged_data[column], ax=axes[i])
    axes[i].set_title(f'{column}')  

# Adjust layout for better spacing
plt.tight_layout()
plt.show()
```

`Amount`'s boxplot shows a right-skewed distribution where a small amount of applicants seek larger amount of loans. There is a significant number of high outliers ranging from `7500 to 16500`, which one outlier very far away from the rest of data at `18000`. The range of high outliers may be due to borrowers applying for large sums for business, education or other investments purposes. It might also be due to wealthier individuals requesting a higher than normal credit amount. This might also be a form of error. 

`Installment` appears to have a high median of `3.0`, with an upper quartile value of `4.0` and lower quartile of `2.0`. The distribution is relatively balanced, with some high end values. There does not seem to have any outliers for this feature.

`Age` shows that majority of the clients are around the late `20s and early 40s`. There are some older borrowers `above 60 years old`. Since there is a considerable amount of data, we cannot assume that these are due to errors in data entry. However, some possble reasons for the outlier values might be due to these senior individuals still being active in business or investments, or might be borrowing in order to fund their medical expenses.

`Num_Credits` is higher tended to the lower values, `mainly 1`. A few individuals have 4 active credits. Either these individuals have the financial capability to manage multiple loans simultaneously, or they are high-risk individuals with multiple credit sources. Since the number of outliers is very small compared to the rest of the data, we will eliminate these outlier values for a fairer analysis. 

```python
# Identify the upper whisker for Amount using IQR method
Q1 = data['Amount'].quantile(0.25)
Q3 = data['Amount'].quantile(0.75)
IQR = Q3 - Q1
upper_whisker = Q3 + 1.5 * IQR

# Get the middle value of the remaining outliers (those still above upper_whisker but ≤ 17500)
outlier_values = data[(data['Amount'] > upper_whisker) & (data['Amount'] < 17500)]['Amount']
middle_outlier_value = outlier_values.median() 

# For the Amount outlier > 17500, impute with the median value of the outlier range
data.loc[data['Amount'] > 17500, 'Amount'] = middle_outlier_value

# Drop rows where Num_Credits == 4
data = data[data['Num_Credits'] != 4]

# Print The Changes 
print(f"Replaced Amount values > 17500 with: {middle_outlier_value}")
print(f"Rows removed (Num_Credits == 4): {data.shape[0] - data[data['Num_Credits'] == 4].shape[0]}")
```

For `Amount`, we decided to impute the data point with `>17500` to a median value of the range of outliers. For `Num_Credits`, we decided to drop the rows with outlier `Num_Credits = 4`. The number of rows dropped is insignificant compared to the scale of our data. 

Since the outliers of `Age` is very closed to the upper quartile, this means 25% of clients fall within the outlier values. Removing these data or imputing them with other values could introduce bias, because the 25% is quite a significant presence in the data.

We decided to perform One Hot Encoding on norminal variables with no intrinsic order. 
These variables include `Purpose`, `Guarantors`, `Housing Type`, `Occupation`, `Personal_Status`, and `Foreign_Worker`.

```python
# Nominal columns that are to be encoded 
nominal_cols = ["Purpose", "Guarantors", "Housing_Type", "Occupation", "Personal_Status", "Foreign_Worker"]

# Apply One Hot Encoding
data_encoded = pd.get_dummies(data, columns=nominal_cols, drop_first=True, dtype=int)

# Get the list of encoded columns (this will include all the newly created one-hot encoded columns)
encoded_cols = [col for col in data_encoded.columns if any(col.startswith(cat) for cat in nominal_cols)]

# Display first few rows of only encoded categorical columns
print("\nSample Encoded Data:\n", data_encoded[encoded_cols].head())

# Print the overall data set after encoding 
print("\nTotal Columns After Encoding:", data_encoded.shape[1])
```

```python
# Adjust grid layout dynamically
num_cols = 3  # Number of columns per row
num_rows = -(-len(categories) // num_cols)  # This ensures we have enough rows

# Create larger figure with multiple rows
fig, axes = plt.subplots(num_rows, num_cols, figsize=(30, 6 * num_rows))  # Adjust height dynamically

# Flatten axes array for easy iteration
axes = axes.flatten()

# Loop through each column and create a pie chart
for i, col in enumerate(categories):
    labels = data[col].map(categories[col])
    
    labels.value_counts().plot(
        kind="pie", autopct="%1.1f%%", startangle=90, cmap="Set3", ax=axes[i], 
        explode=[0.05] * labels.nunique()  # Slight separation for better visibility
    )
    axes[i].set_title(f"Distribution of {col}")
    axes[i].set_ylabel("")  # Hide y-axis label for clarity

# Hide unused subplots if categories don't fill all slots
for j in range(i + 1, len(axes)):
    fig.delaxes(axes[j])

# Adjust layout for better visibility
plt.tight_layout()
plt.show()
```
